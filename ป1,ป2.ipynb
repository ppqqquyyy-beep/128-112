```python
#ป.1 ชุดฟังก์ชันตรวจข้อมูล
def is_valid_score(score):
  return 0 <= score <= 100

def to_grade(score):
  if score >= 80:
    return "A"
  elif score >= 70:
    return "B"
  elif score >= 60:
    return "C"
  else:
    return "F"

def print_result(student_id, score):
    if is_valid_score(score):
      grade = to_grade(score)
      print(f"รหัสนักศึกษา: {student_id} คะแนน: {score} เกรด: {grade}")
    else:
      print(f"รหัสนักศึกษา: {student_id} คะแนน: {score} ข้อความแจ้งเตือน: คะแนนไม่ถูกต้อง (ต้องอยู่ระหว่าง 0 ถึง 100)")
print_result("65010001", 85)
```

```python
#ป.2 สถิติแบบไหลผ่าน
def safe_average(total, count):
  if count <= 0:
    return None

    average = total/count
    return round(average, 2)

day1 = float(input("กรอกค่า PM2.5 วันที่ 1: "))
day2 = float(input("กรอกค่า PM2.5 วันที่ 2: "))
day3 = float(input("กรอกค่า PM2.5 วันที่ 3: "))
day4 = float(input("กรอกค่า PM2.5 วันที่ 4: "))
day5 = float(input("กรอกค่า PM2.5 วันที่ 5: "))
total_pm25 = 0
count_days = 0

for day in [day1, day2, day3, day4, day5]:
  if day >= 0:
    total_pm25 += day
    count_days += 1

result = safe_average(total_pm25, count_days)

if result is not None:
  print(f"ค่าเฉลี่ย PM2.5 (2 ตำแหน่ง): {result}")
else:
  print("ไม่มีข้อมูลที่ใช้ได้ (count ไม่เป็นบวก)")
```
