```python
# ป.1 สรุปคะแนน
scores=[78, 92, 65, 88, 45, 95, 72]
count=len(scores)
print(f"จำนวนผู้เข้าสอบ {count} คน")
print(f"คะแนนสูงสุด {max(scores)} คะแนน")
print(f"คะแนนต่ำสุด {min(scores)} คะแนน")
total_score=sum(scores)
average=total_score/count
average2=round(average, 2)
print(f"ค่าเฉลี่ย {average2}")
a=[]
f=[]
for x in scores:
  if (x >= 80):
    a.append(x)
    student80=len(a)
  elif (x < 50):
    f.append(x)
print(f"จำนวนผู้ได้ 80 คะแนนขึ้นไป: {student80} คน")
print(f"คะแนนที่ไม่ผ่านเกณฑ์ 50: {f}")
```

```python
# ป.2 ทำความสะอาดข้อมูลซ้ำ
ids = ["C001", "C002", "C001", "C003", "C002", "C004"]
number=len(ids)
print(f"จำนวนรายการทั้งหมด : {number}")
unique_code=len(list(set(ids)))
print(f"จำนวนรหัสผ่านที่ไม่ซ้ำ : {unique_code}")
seen=set()
duplicates=set()
for x in ids:
  if x in seen:
    duplicates.add(x)
  else:
    seen.add(x)
print(f"จำนวนรายการที่ซ้ำ : {len(list(duplicates))}")
print(f"รหัสที่ไม่ซ้ำ : {list(dict.fromkeys(ids))}")
```

```python
# ป.3 นับความถี่และหาค่าสูงสุด
payments = ["cash", "credit", "cash", "transfer", "cash", "credit"]
count_list={}
for x in payments:
  count_list[x]=count_list.get(x,0)+1
print(count_list)
print(f"วิธีที่ใช้มากที่สุด {max(count_list, key=count_list.get)}")
```

```python
# ป.4 รายงานจากตารางข้อมูล
orders = [
{"id": 1001, "customer": "Alice", "city": "Bangkok", "product": "Notebook", "qty": 2, "price":
120},
{"id": 1002, "customer": "Bob", "city": "Bangkok", "product": "Pen", "qty": 5, "price": 15},
{"id": 1003, "customer": "Charlie", "city": "Chiang Mai", "product": "Backpack", "qty": 1, "price":
850},
{"id": 1004, "customer": "Diana", "city": "Phuket", "product": "Notebook", "qty": 3, "price":
120},
{"id": 1005, "customer": "Eve", "city": "Bangkok", "product": "Pen", "qty": 10, "price": 15},
{"id": 1006, "customer": "Frank", "city": "Phuket", "product": "Backpack", "qty": 1, "price": 850},
]
total_sales=0
product_sales={}

for x in orders:
  subtotal=x["qty"]*x["price"]
  net_price=subtotal
  total_sales+=net_price
  product_sales[x["product"]] = x["qty"]
  unique_customers={x["customer"] for x in orders}


print(f"ยอดขายรวมทั้งหมด : {total_sales} บาท")
print(product_sales)
print(f"จำนวนลูกค้าที่ไม่ซ้ำ : {len(unique_customers)} คน")

top_customers=sorted(
    customer_sales.items(),
    key=lambda x: x[1],
    reverse=True
)[:3]

print("ลูกค้า 3 อันดับแรกที่มียอดขายสูงสุด")
print(top_customers)
```

```python
# ป.5 เลือกโครงสร้างด้วยตนเอง
#โครตงสร้างที่เลือก Dictionary
attendance = {
    "A": [1, 2, 3],
    "B": [1, 3],
    "C": [2, 4],
    "D": [],
    "E": [1, 2, 3, 4]
}
never_attended = []

for student, weeks in attendance.items():
    if len(weeks) == 0:
        never_attended.append(student)

print("นักศึกษาที่ไม่เคยเข้าเรียนเลย:")
print(never_attended)
#เหตุผลที่เลือกใช้โครงสร้างนี้ Dictionary สามารถจัดเก็บข้อมูลโดยใช้ชื่อนักศึกษาเป็น Key ทำให้ค้นหาข้อมูลได้ง่าย
```
