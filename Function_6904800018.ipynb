```python
#2.สี่เหลี่ยมผืนผ้า
def rectangle_area(width, height): #คืนค่าพื้นที่
  return width*height

def rectangle_perimeter(width, height): #คืนค่าเส้นรอบรูป
  return 2*(width+height)

def show_rectangle(width, height):
  print(f"กว้าง {width} ยาว {height}")
  print(f"พื้นที่ : {rectangle_area(width, height)}")
  print(f"เส้นรอบรูป : {rectangle_perimeter(width, height)}")
show_rectangle(10, 10)

#ถ้าเขียน x = show_rectangle(5, 3) แล้ว print(x) จะได้อะไร เพราะเหตุใด
#ผลลัพธ์ุที่ได้คือ None เพราะไม่มี return
```

```python
#3.เลขคู่
def is_even(number):
  return number % 2 == 0 #คู่
print(is_even(8))
print(is_even(7))
print(is_even(0))

def count_even(start, end):
  count = 0
  for x in range(start, end+1):
    if is_even(x):
      count += 1
  return count
print(count_even(1, 20))
print(count_even(15, 31))
```

```python
#4.ค่ามากที่สุดจากสามจํานวน
def max_of_three(a, b, c):
  max_val = a
  for num in a, b, c:
    if num > max_val:
      max_val = num
  return max_val
print(max_of_three(3, 9, 5))
print(max_of_three(9, 3, 5))
print(max_of_three(3, 5, 9))
print(max_of_three(7, 7, 2))
print(max_of_three(-1, -8, -3))

print("---------------")

def min_of_three(a, b, c):
  min_val = a
  for num in a, b, c:
    if num < min_val:
      min_val = num
  return min_val
print(min_of_three(3, 9, 5))
print(min_of_three(9, 3, 5))
print(min_of_three(3, 5, 9))
print(min_of_three(7, 7, 2))
print(min_of_three(-1, -8, -3))

print("---------------")

def range_of_three(a, b, c):
  max_val = max_of_three(a, b, c)
  min_val = min_of_three(a, b, c)
  return max_val - min_val
print(range_of_three(3, 9, 5))
print(range_of_three(9, 3, 5))
print(range_of_three(3, 5, 9))
print(range_of_three(7, 7, 2))
print(range_of_three(-1, -8, -3))
```

```python
#5.สรุปคะแนนจาก list และคืนค่าเป็น tuple
def score_summary(scores):
  if not scores:
    return None

  min_score = scores[0]
  max_score = scores[0]
  avg_score = scores[0]
  total = 0

  for num in scores:
    if num < min_score:
      min_score = num
    elif num > max_score:
      max_score = num
  for num in scores:
    total += num/len(scores)
    average_score = total

  return min_score, max_score, round(average_score, 2)


print(score_summary([80, 65, 90, 72, 58]))
print(score_summary([50]))
print(score_summary([]))

#ถ้าแตก tuple จากค่า None โดยไม่ตรวจก่อน จะเกิด error อะไร
#TypeError เพราะประเภทข้อมูลไม่รองรับการวนซ้ำ
```

```python
#7.ผลรวมเลขโดดและจํานวนเฉพาะ
def sum_digits(number):
  total = 0
  while number > 0:
    digits = number % 10
    total += digits
    number = number // 10
  return total
print(sum_digits(2569))
print(sum_digits(1000))
print(sum_digits(0))

def is_prime(number):
  if number <= 1:
    return False
  for i in range(2, int(number**0.5) + 1):
    if number % i == 0:
      return False
  return True
print(is_prime(1))
print(is_prime(2))
print(is_prime(91))
print(is_prime(97))

def list_primes(start, end):
  primes = []
  for num in range(start, end + 1):
    if is_prime(num):
      primes.append(num)
  return primes
print(list_primes(1, 20))
print(list_primes(90, 100))
print(list_primes(24, 28))

def count_primes(start, end):
  return len(list_primes(start, end))
print(count_primes(1, 50))
```

```python
#8.วิเคราะห์ผลโหวตด้วย set และ dict พร้อมชุดทดสอบ

votes = ["มะม่วง", "ส้ม", "มะม่วง", "กล้วย", "ส้ม", "มะม่วง"]
def unique_items(items):
  return set(items)

def count_frequency(items):
  freq = {}
  for item in items:
    if item in freq:
      freq[item] += 1
    else:
      freq[item] = 1
  return freq

def most_common(items):
  if not items:
    return None
  freq = count_frequency(items)
  max_item = None
  max_count = 0
  for item, count in freq.items():
    if count > max_count:
      max_item = item
      max_count = count
  return (max_item, max_count)

def test_votes():
  assert unique_items(votes) == {"มะม่วง", "ส้ม", "กล้วย"}
  assert unique_items([]) == set() #listว่าง
  assert count_frequency(votes) == {"มะม่วง": 3, "ส้ม": 2, "กล้วย": 1}
  assert count_frequency([]) == {} #listว่าง
  assert most_common(votes) == ("มะม่วง", 3)
  tie_votes = ["ส้ม", "กล้วย", "ส้ม", "กล้วย"]
  assert most_common(tie_votes) == ("ส้ม", 2) #กรณีจำนวนเท่ากัน
  assert most_common(["ส้ม"]) == ("ส้ม", 1) #กรณีมีรายการเดียว
  assert most_common([]) is None #listว่าง

print(unique_items(votes))
print(len(unique_items(votes)))
print(count_frequency(votes))
print(most_common(votes))
print(most_common(["ส้ม", "กล้วย"]))
print(unique_items([]))
print(count_frequency([]))
print(most_common([]))

#set ไม่สนลำดับมีข้อมูลเหมือนกันก็เท่ากันแล้ว list สนลำดับถ้าเรียงไม่ตรงกันจะไม่เท่ากัน
```

```python
#9.เรียกซ้ําเทียบกับวนซ้ํา

def power_loop(base, exponent):
  result = 1
  for _ in range(exponent):
     result *= base
  return result

def power_recursive(base, exponent):
  if exponent == 0:
    return 1
  return base

def gcd_loop(a, b):
  while b != 0:
    a, b = b, a % b
  return a

def gcd_recursive(a, b):
  if b == 0:
    return a
  return gcd_recursive(b, a % b)

def sum_list_loop(numbers):
  total = 0
  for num in numbers:
    total += num
  return total

def sum_list_recursive(numbers):
  if not numbers:
    return 0
  return numbers[0]

print(power_loop(2, 10))
print(power_loop(5, 0))
print(gcd_loop(48, 18))
print(gcd_loop(17, 5))
print(gcd_loop(100, 75))
print(gcd_loop(12, 0))
print(sum_list_loop([4, 8, 15]))
print(sum_list_loop([]))
```
